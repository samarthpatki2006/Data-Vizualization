```python
df['Toss_Match_Winner']=df['toss_winner']==df['winner']
```

```python
df11=df['Toss_Match_Winner'].value_counts().reset_index(name='Count')
```

```python
df11['Toss_Match_Winner']=df11['Toss_Match_Winner'].apply(lambda x: 1 if x==True else 0)
```

```python
df11
```

```python
plt.pie(df11['Count'],labels=['Toss Winner Wins Match','Toss winner looses match'],autopct='%1.1f%%')
plt.title('Winner vs Toss Winner')
```
