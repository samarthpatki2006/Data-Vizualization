```python
def replaceNoisy(col):
  q1=col.quantile(0.25)
  q3=col.quantile(0.75)
  iqr=q3-q1
  lower=q1-(1.5*iqr)
  upper=q3*(1.5*iqr)
  median=col.median()
  return col.apply(lambda x: median if(x<=lower) |(x>=upper) else x)
df2[df2.select_dtypes(include='number').columns]=df2.select_dtypes(include='number').apply(replaceNoisy)
```
