```python
df1=df['winner'].value_counts().reset_index(name='count')
```

```python
df1
```

```python
df1.info()
```

```python
plt.figure(figsize=(12,6))
plt.bar(df1['winner'],df1['count'],width=0.7)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.barplot(x=df1['winner'],y=df1['count'],hue=df1['winner'],palette='viridis')
plt.xticks(rotation=90)
```
