BOLLYWOOD DATASET

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../Datasets/bollywood.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df['Genre']=df['Genre'].str.strip()
```

```python
df1=df['Genre'].value_counts().reset_index(name='Count')
```

```python
df1
```

```python
sns.barplot(x=df1['Genre'],y=df1['Count'],hue=df1['Count'],palette='magma')

plt.show()
```

```python
df1[df1['Count']==df1['Count'].max()]['Genre']
```

```python
df1=df.groupby('Genre')['ReleaseTime'].value_counts().reset_index(name='Count')
```

```python
df1
```

```python
df1=df['ReleaseTime'].value_counts().reset_index(name='Count')
```

```python
df1
```

```python
sns.barplot(x=df1['ReleaseTime'],y=df1['Count'],palette='magma')
plt.show()
```

```python
df1[df1['Count']==df1['Count'].max()]['ReleaseTime']
```

```python
df1=df[df['Budget']>25]
```

```python
df2=df1.groupby('ReleaseTime')['MovieName'].value_counts().reset_index(name='Count')
```

```python
df3=df2.groupby('ReleaseTime')['Count'].sum().reset_index(name='Count')
```

```python
df3
```

```python
sns.barplot(x=df3['ReleaseTime'],y=df3['Count'])
plt.show()
```

```python
df['Profit']=df['BoxOfficeCollection']-df['Budget']
```

```python
df1=df.sort_values(by='Profit',ascending=False)
```

```python
df1=df1[['MovieName','Profit']].head(10)
```

```python
sns.barplot(x=df1['MovieName'],y=df1['Profit'],palette='viridis')
plt.xticks(rotation=90)
plt.show()
```

```python
df1=df.groupby('ReleaseTime')['Profit'].mean().reset_index(name='AvgProfit')
```

```python
df1
```

```python
sns.barplot(x=df1['ReleaseTime'],y=df1['AvgProfit'],palette='viridis')
plt.show()
```

```python
df1=df.groupby('Genre')['YoutubeLikes'].sum().reset_index(name='Total no of Likes')
```

```python
df1
```

```python
sns.barplot(x=df1['Genre'],y=df1['Total no of Likes'],palette='viridis')
plt.show()
```
