```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data1={'A':['a','b','c'],'B':[1,2,3]}
df1=pd.DataFrame(data1)
```

```python
data2={'A':['a','b','c','a','c','a','b'],'C':[100,200,300,100,200,100,200]}
df2=pd.DataFrame(data2)
```

```python
df3=df2.merge(df1,on='A')
```

```python
df3
```

```python
df3.plot(kind='bar',stacked=True)
```

```python
data={'A':[10,20,30],'B':['Category1','Category2','Category3']}
df=pd.DataFrame(data)
```

```python
plt.pie(df['A'],labels=df['B'],startangle=190,explode=[0.05,0.05,0],colors=sns.color_palette('magma'),autopct='%1.1f%%',shadow=True)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8,5))
ax.pie(df['A'],labels=df['B'],startangle=190,explode=[0,0.05,0],colors=sns.color_palette('magma'),autopct='%1.1f%%',shadow=True)
ax.set_aspect(0.6)
plt.show()
```
