```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, np.nan, 30, 45, np.nan],
    'City': ['New York', 'Los Angeles', np.nan, 'Chicago', 'New York'],
    'Salary': [70000, 80000, np.nan, 120000, 95000]
}
df=pd.DataFrame(data)
df.to_csv('Sample.csv',index=False)
```

```python
df
```

```python
df.isna()
```

```python
df.isnull().sum()
```
