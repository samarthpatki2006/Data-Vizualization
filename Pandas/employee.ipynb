EMPLOYEE DATASET

```python
import pandas as pd
```

```python
data={
  'Id':[101,102,103,104,105,106,107,108],
  'Name':['Alice','Bob','Charlie','David','Emma','Frank','Grace','Helen'],
  'Department':['HR','IT','IT','Finance','HR','IT','Finance','HR'],
  'Salary':[60000,75000,80000,72000,65000,78000,70000,62000],
  'ExperienceYears':[5,7,6,8,4,9,6,3]
}
df=pd.DataFrame(data)
```

```python
df.to_csv('Employee.csv',index=False)
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Department'].nunique()
```

```python
print('Employee with max salary: ')
maxSalary=df['Salary'].max()
# df[df['Salary']==maxSalary][['Name','Salary']] ----alternate method
df.loc[df['Salary']==maxSalary,['Name','Salary']]
```

```python
print("Employees working in IT department: ")
df[(df['Department']=='IT') & (df['Salary']<80000)][['Name','Salary','Department']]
```

```python
print('Department wise max salary')
df1=df.groupby('Department')['Salary'].max().reset_index()
df1
```

```python
print("Department wise details: ")
df1=df.groupby('Department').agg({'Salary':['min','max','mean'],'ExperienceYears':['sum']}).reset_index()
df1
```

```python
df1=df.groupby(['Department','ExperienceYears'])['Salary'].mean().reset_index()
df1
```

```python
print('Department wise employee count: ')
# df1=df.groupby('Department')['Id'].count().reset_index(name='Count')
# df1
df.value_counts('Department').reset_index(name='No of employees')
```

```python
df1=df.sort_values(by=['Salary','ExperienceYears'],ascending=[False,True])
df1
```

```python
df['Tax']=df['Salary']*0.30
df
```
