```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=df.replace('?',np.nan)
df
```

```python
df.info()
```

```python
df[['carbo','sugars','potass']]=df[['carbo','sugars','potass']].apply(pd.to_numeric)
```

```python
df.info()
```

```python
df=df.fillna(df.mean(numeric_only=True))
```

```python
df.info()
```

```python
print('NO of manufacturers: ')
df['mfr'].nunique()
```

```python
d1=df.groupby('mfr')['name'].count().reset_index(name='No of cereals')
d1
```

```python
plt.bar(d1['mfr'],d1['No of cereals'])
plt.title('No of Cereals per Manufacturers')
plt.xlabel('Manufacturers')
plt.ylabel('No of cereals')
plt.show()
```

```python
sns.barplot(x=d1.mfr,y=d1['No of cereals'],hue=d1.mfr,palette='viridis')
```
