```python
# fill na values in numeric cols
df=df.fillna(df.mean(numeric_only=True))
```

```python
# fill na in objects column
df=df.fillna(df.mode().iloc[0])
```

```python
df.isna().sum()
```

```python
q1=df['BMI'].quantile(0.25)
q1
```

```python
df.head()
```

```python
# handling outliers
def replaceNoisy(col):
  q1=col.quantile(0.25)
  q3=col.quantile(0.75)
  return col.apply(lambda x: col.median() if((x>q3) or (x<q1)) else x)
df[df.select_dtypes(include='number').columns]=df.select_dtypes(include='number').apply(replaceNoisy,axis=0)
```
