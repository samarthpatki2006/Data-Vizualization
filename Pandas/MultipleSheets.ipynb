Create an excel file called StdMrks.xls with two sheets. 1 sheet contains roll number, name of 5 students with their marks in 5 quizzes (out of 10). The other sheet contains marks obtained in two exams (out of 100 each ) for the same student.

```python
# Reading the file , sheet by sheet
xlsx = pd.ExcelFile("StudMrks.xlsx")
```

```python
# This object can show you the list of available sheet names in the file and how to parse them 
xlsx.sheet_names
```

```python
df_mrks1=xlsx.parse(sheet_name="Quiz")
df_mrks2=xlsx.parse(sheet_name="Exam", usecols=["E1", "E2"])
```

```python
# f you are reading multiple sheets in a file, then can also simply pass the filename to pandas.read_excel:
frame1 = pd.read_excel("StudMrks.xlsx", sheet_name="Quiz")
frame2 =  pd.read_excel("StudMrks.xlsx", sheet_name="Exam",usecols=["E1", "E2"])
#frame1
frame2
# Combine both DataFrames side by side (column-wise merge)
combined_df = pd.concat([frame1, frame2], axis=1)

# Save the combined data to a new Excel file (optional)
combined_df.to_excel("combined_dataset.xlsx", index=False)

# Display the first few rows
print(combined_df.head())
```

```python
import pandas as pd

# Load the Excel file
file_path = "SMnew.xlsx"  # Replace with the actual file path

# Specify the sheets and the columns to extract
sheets_to_combine = { "Quiz", "Exam": ["E1", "E2"]  # Columns C and D from Sheet2
}

# Read only the selected columns from the specified sheets
sheet_dfs = {sheet: pd.read_excel(file_path, sheet_name="Quiz", usecols=columns) 
             for sheet, columns in sheets_to_combine.items()}

# Merge both DataFrames
combined_df = pd.concat(sheet_dfs.values(), axis=1)  # Merging side by side

# Save the new dataset to a new Excel file (optional)
combined_df.to_excel("combined_selected_columns.xlsx", index=False)

# Display the first few rows
print(combined_df.head())
```

```python
import pandas as pd

# Load the Excel file
file_path = "StudMrks.xlsx"  # Replace with your actual file path

# Read ALL columns from "Quiz" sheet (Sheet1)
df1 = pd.read_excel(file_path, sheet_name="Quiz")  # Reads all columns

# Read ONLY specific columns from "Exam" sheet (Sheet2)
df2 = pd.read_excel(file_path, sheet_name="Exam", usecols=["E1", "E2"])  # Replace with actual column names

# Combine both DataFrames side by side (column-wise merge)
combined_df = pd.concat([df1, df2], axis=1)

# Save the combined data to a new Excel file (optional)
combined_df.to_excel("combined_dataset.xlsx", index=False)

# Display the first few rows
print(combined_df.head())
```

```python
import pandas as pd

# Load the Excel file
file_path = "SMnew.xlsx"  # Replace with your actual file path

# Read specific columns from "Quiz" sheet
frame1 = pd.read_excel(file_path, sheet_name="Quiz", usecols=["ColumnA", "ColumnB", "ColumnC"])  # Replace with actual column names

# Read specific columns from another sheet, e.g., "Exam"
frame2 = pd.read_excel(file_path, sheet_name="Exam", usecols=["ColumnX", "ColumnY"])  # Replace with actual column names

# Combine both DataFrames side by side (column-wise merge)
combined_df = pd.concat([frame1, frame2], axis=1)

# Save the combined data to a new Excel file (optional)
combined_df.to_excel("combined_dataset.xlsx", index=False)

# Display the first few rows
print(combined_df.head())
```
