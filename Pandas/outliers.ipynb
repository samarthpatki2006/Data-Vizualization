```python
import pandas as pd
```

```python
data={'A':[1,2,3,10000,10],'B':['a','b','c','c','e']}
```

```python
df=pd.DataFrame(data)
```

```python
def replaceNoisy(col):
  q1=col.quantile(0.25)
  q3=col.quantile(0.75)
  iqr=q3-q1
  lower=q1-(1.5*iqr)
  upper=q3+(1.5*iqr)
  m=col.median()
  return col.apply(lambda x: m if(x<lower) | (x>upper) else x)
df[df.select_dtypes(include='number').columns]=df.select_dtypes(include='number').apply(replaceNoisy)
```
